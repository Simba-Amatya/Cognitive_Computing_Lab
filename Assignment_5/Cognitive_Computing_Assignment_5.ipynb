```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)

print("After adding 2:", arr + 2)

print("After multiplying by 3:", arr * 3)

print("After dividing by 2:", arr / 2)
```

```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print("\nArray x:", x)
print("Most frequent value:", most_frequent)
print("Indices:", indices)

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]

print("\nArray y:", y)
print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
import numpy as np

Amatya_Simba = np.linspace(10, 100, 25)

print("Array:")
print(Amatya_Simba)

print("Dimensions:", Amatya_Simba.ndim)
print("Shape:", Amatya_Simba.shape)
print("Total elements:", Amatya_Simba.size)
print("Data type:", Amatya_Simba.dtype)
print("Total bytes:", Amatya_Simba.nbytes)

transpose_array = Amatya_Simba.reshape(1, 25)

print("Transpose using reshape():")
print(transpose_array)

transpose_T = Amatya_Simba.T

print("Transpose using T attribute:")
print(transpose_T)

print("Can we do the same with T attribute? Yes, but for a 1-D array, T does not change its shape.")
```

```python
import numpy as np

ucs420_Amatya_Simba = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 25, 35]
])

print("Original array:")
print(ucs420_Amatya_Simba)

print("Mean:", np.mean(ucs420_Amatya_Simba))

print("Median:", np.median(ucs420_Amatya_Simba))

print("Maximum:", np.max(ucs420_Amatya_Simba))

print("Minimum:", np.min(ucs420_Amatya_Simba))

print("Unique elements:", np.unique(ucs420_Amatya_Simba))

reshaped_ucs420_Amatya_Simba = ucs420_Amatya_Simba.reshape(4, 3)

print("\nReshaped array (4 x 3):")
print(reshaped_ucs420_Amatya_Simba)

resized_ucs420_Amatya_Simba = np.resize(ucs420_Amatya_Simba, (2, 3))

print("\nResized array (2 x 3):")
print(resized_ucs420_Amatya_Simba)
```
